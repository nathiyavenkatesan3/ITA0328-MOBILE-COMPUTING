```python
import numpy as np
np.random.seed(1)
N_UE, N_RB, N_TTI = 4, 10, 1000
base = np.array([10, 7, 4, 1.5])          # avg Mbps per RB (UE0 near ... UE3 far)

def jain(x):
    return x.sum()**2 / (len(x) * (x**2).sum())

def run(policy):
    total = np.zeros(N_UE)
    avg = np.ones(N_UE)                   # PF average throughput
    for t in range(N_TTI):
        rate = base * np.random.exponential(1, N_UE)   # fading this TTI
        got = np.zeros(N_UE)
        for rb in range(N_RB):
            if policy == "MaxCI":
                u = np.argmax(rate)
            elif policy == "RR":
                u = (rb + t) % N_UE
            else:                         # PF
                u = np.argmax(rate / avg)
            got[u] += rate[u]
        avg = 0.99 * avg + 0.01 * got
        total += got
    return total / N_TTI

for p in ["MaxCI", "RR", "PF"]:
    x = run(p)
    print(f"{p:6} per-UE = {np.round(x, 1)} Total = {x.sum():.1f} Jain = {jain(x):.3f}")
```
