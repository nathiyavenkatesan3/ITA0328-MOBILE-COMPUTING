```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 1001, 10)                # mobile moves from BS1 (0 m) to BS2 (1000 m)
p1 = -30 - 35*np.log10(np.maximum(x, 1))          # signal from BS1 (dBm)
p2 = -30 - 35*np.log10(np.maximum(1000 - x, 1))   # signal from BS2 (dBm)
H, serving = 5, 1                         # hysteresis margin = 5 dB

for i in range(len(x)):
    if serving == 1 and p2[i] > p1[i] + H:
        serving = 2
        print(f"Handoff BS1 -> BS2 at {x[i]} m")
        break

plt.plot(x, p1, label="BS1"); plt.plot(x, p2, label="BS2")
plt.xlabel("Distance (m)"); plt.ylabel("Signal (dBm)"); plt.legend(); plt.show()
```
