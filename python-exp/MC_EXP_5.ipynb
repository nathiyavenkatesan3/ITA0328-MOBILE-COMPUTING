```python
import numpy as np

code1 = np.array([1, -1, 1, -1])   # user 1 code
code2 = np.array([1, 1, -1, -1])   # user 2 code
bit1 = 1                           # user 1 data
bit2 = -1                          # user 2 data

tx = bit1 * code1 + bit2 * code2   # both signals add in the channel
print("Sent signal      :", tx)
print("User 1 recovered :", np.sign(tx @ code1 / 4))
print("User 2 recovered :", np.sign(tx @ code2 / 4))
```
